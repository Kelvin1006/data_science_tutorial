```python
import numpy as np
```

```python
#1.Create a numpy array of size 2 x 2
array2x2=np.array([[1,0],[0,6]])
print(array2x2)
```

```python
#2.Create a random numpy array of size 3 x 4
randomarray3x4 = np.random.rand(3, 4)
print(randomarray3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes
arrayofones = np.ones((10, 6))
print("Numpy array of ones:")
print(arrayofones)

arrayofzeroes = np.zeros((6, 7))
print("Numpy array of zeroes:")
print(arrayofzeroes)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it
randomarray6x3 = np.random.randint(0,10,size=(6, 3))
print(randomarray6x3)
transposedarray = randomarray6x3.T
print("Transposed array:")
print(transposedarray)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product
n,p,m=3,4,3
randomarray1 = np.random.rand(n, p)
randomarray2 = np.random.rand(p, m)
print("Random array 1:")
print(randomarray1)
print("Random array 2:")
print(randomarray2)

dotproduct = np.dot(randomarray1, randomarray2)
print("Dot Product:")
print(dotproduct)
```
